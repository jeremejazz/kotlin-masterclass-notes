# Check Instructions

Check instructions are mechanisms used to validate certain conditions in your code, typically to ensure that function arguments meet specific constraints. They are advantageous because they simplify error handling and make the code easier to read and maintain.

Check instructions are often employed at the beginning of functions to validate input parameters. They can also appear at the end of functions to ensure that the results are as expected.

````kotlin
fun average(value: Int){
    require(value > 0){
        "Invalid Number"
    }

}

average(0)
````

````kotlin

fun multiplyWithTen(number: Int?): Int{
    requireNotNull(number){
        "We need a nmot null number"
    }

    return number * 10
}

multiplyWithTen(null)
````
