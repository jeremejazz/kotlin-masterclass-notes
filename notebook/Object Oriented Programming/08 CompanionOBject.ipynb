# Companion Object

- All functions inside companion object are all about class
- companion object elements cannot access regular class elements

```kotlin

class Example {
    fun callMe() = "Calling a function"
    val result = callMe2()
    companion object{
        fun callMe2() = "Calling a function."
        val result = this
    }
}


val example = Example()
example.result

```

- we can only have 1 companion object within a class
- companion objects also acts as a singleton object
