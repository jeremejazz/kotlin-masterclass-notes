# Upcasting and Downcasting



```kotlin
interface Animal{
    fun eat()
    fun run()
}

class Dog: Animal{
    override fun eat(){
        println("${this::class.qualifiedName} is eating...")
    }

    override fun run() {
        println("${this::class.qualifiedName} is running...")
    }

    fun fetchABall(): Unit {
    println("Fetching a Ball")
    }
}

class Cat: Animal{
    override fun eat() {
        println("${this::class.qualifiedName} is eating...")
    }

    override fun run() {
        println("${this::class.qualifiedName} is running...")
    }

    fun scratch(): Unit {
        println("${this::class.qualifiedName} scratch...")
    }

}
// option 1 using if
fun getAnimal(animal: Animal){
    animal.eat()
    animal.run()

    if(animal is Dog){ // smart cast or automatic downcast
        animal.fetchABall()
    }
}

// using when
fun getAnimal2(animal: Animal){
    when(animal){
        is Dog -> {
            animal.fetchABall()
        }
        is Cat -> {
            animal.scratch()
        }
    }
}

// using as keyword

fun getAnimal3(animal: Animal): Unit {
    // unsafe cast
    val dog  = animal as Dog
    dog.fetchABall()
}


fun getAnimal4(animal: Animal): Unit {
    // alternative that uses nullable
    val dog  = animal as? Dog
    dog?.fetchABall()
}

val dog = Dog()
val cat = Cat()
getAnimal(animal = dog)
```
