# Sealed Classes
A sealed class in Kotlin is a special type of class that restricts inheritance.
- Sealed classes are abstract classes that can only be subclassed within the same file where they are declared.
- They are used to represent restricted class hierarchies, where a value can be one of the predefined types. This helps in exhaustive when expressions, making the code more safe and maintainable.
- Typically, sealed classes are used in scenarios like state management, result handling, etc., where the type can be one of a limited set of options.

```kotlin
sealed class Result
class Success(val data: String) : Result()
class Error(val exception: Throwable) : Result()
```
