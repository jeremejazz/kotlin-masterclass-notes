# Object

- represents a singleton. Allowed to only have a single instance in the application

````kotlin

object SingleTon{
    fun printName(){
        println("Hi, I am Ted")
    }
}

SingleTon.printName()
````

Visibility modifiers are applicable

```kotlin
private object SingleTon{
    fun printName(){
        println("Hi, I am Ted")
    }
}
```

````kotlin
object Counter{
    private var count = 0
    fun increment(){
        ++count
    }
    fun currentCount() = count
}

Counter.increment()
Counter.increment()
println(Counter.currentCount())


 
````

- Objects also support Inheritance
- Object cannot be placed inside a function but can be in another object.
