# Nested Classes

Nested class is a class within the scope of an outer class

```kotlin
class Airport(private val code: String){
    open class Plane{
        fun contact(airport: Airport){
            println(airport.code)
        }
    }

    private class PrivatePlane: Plane(){
        fun secretCode(){
            println("Secret Code")
        }
    }
}


val airport = Airport(code="ENG")
val plane = Airport.Plane()
plane.contact(airport = airport)
```
