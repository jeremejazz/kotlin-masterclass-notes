## Ineritance

- Base classes require an open keyword first. Same applies to member functions to override
- derived class should initialize the base class.
- parent class can have extension function, the child class can also inherit them.

```kotlin
open class Animal(val name: String){
    protected var energy = 0
    fun printEnergy(){
        println("Current name: $name Energy: $energy")
    }
}

class Dog (dogName: String): Animal(name = dogName){
    fun eat(){
        energy++
    }
}

val d = Dog("Brownie")
d.printEnergy()
d.eat()
d.printEnergy()
```
