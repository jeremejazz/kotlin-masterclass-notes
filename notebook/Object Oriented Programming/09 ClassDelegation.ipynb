# Class Delegation

````kotlin
interface Programmer{
    var working: Boolean
    fun work()
    fun getPaid()
}

class KotlinProgrammer: Programmer{
    override var working = false

    override fun work(){
        println("Programming with Kotlin...")
        working = true
    }

    override fun getPaid() {
        if(working) {
            println("Getting the Money")
        }else{
            println("Need to work first")
        }
        working = false
    }
}

// no need to override programmer because already delegated
class Freelancer: Programmer by KotlinProgrammer()
````

You cannot delegate 2 different classes that uses the same interface
