# Composition

Kotlin encourages the use of composition over inheritance

```kotlin
interface Animal{
    fun eat()
    fun run()
}

open class Dog: Animal{
    override fun eat(){}
    override fun run(){}
    open fun bark(){}
}

class Cat: Animal{
    override fun eat(){}
    override fun run(){}
    fun hiss(){}
}

interface Robot{
    fun move()
}

class CleaningRobot: Robot {
    fun clean(){
    }

    override fun move() {

    }
}
//class CleaningRobotDog: Animal, Robot, Dog() {
//    override fun move() {
//        TODO("Not yet implemented")
//    }
//
//}

/// composition
class CleaningRobotDog{
    private val dog = Dog()
    private val cleaningRobot = CleaningRobot()

    fun run(){
        dog.run()
    }

    fun move(){
        cleaningRobot.move()
    }

    fun bark(){
        dog.bark()
    }
}

// can also be injected
```
