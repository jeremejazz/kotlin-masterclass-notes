# Lambdas

```kotlin
//

val numbers = listOf(1,2,3)
val quotedNumbers = numbers.map{
    number -> "\"$number\""
}

println(quotedNumbers)
```

Rules:
- Lambda expression is always surrounded by curly brackets
- the parameters inside a lambda can have optional type annotation
- The actual lambda body goes after the arrow symbol `->`
- If the returned type of the lambda is not a `Unit` then the last value is treated as a return value

```kotlin
// we can also pass lambda as parameter

val numbers = listOf(1,2,3)
val quotedNumbers = numbers.map( transform = {   number -> "\"$number\""} )
println(quotedNumbers)

// if lambda is at the end of the parameter list, there is no need to call the other parameters when invoking
val quotedNumbers2 = numbers.joinToString { number -> "\"$number\"" }

println(quotedNumbers2)

// type needs to be specified for lambda declaration
val quoted = { number: Int -> "\"$number\""  }
val quotedNumbers3 = numbers.joinToString(transform = quoted)
println(quotedNumbers2)
```

## Multiple parameters in lambda

```kotlin

val quotedNumbers4 = numbers.mapIndexed { index, i -> "\"$i\""  } // can also replace index with _
println(quotedNumbers4)
```
