# Working With Maps

```kotlin

data class Person(
    val name: String,
    val age: Int
)

val names = listOf("Alice", "Bob", "Bill", "Marry", "John")

val ages = listOf(21, 15, 25, 25, 31)

val people = names.zip(ages) { name, age ->
    Person(name, age)
}

println(people)

println("Group people by age Property:")
val map = people.groupBy(Person::age) // :Map<Int, List<Person>>

println(map)
```

## Associating


```kotlin
// In the bottom example, the value becomes the key
val map2 = people.associateWith(Person::name)
println(map2)

println("associate by provided key:")
val map3 = people.associateBy(Person::name)
println(map3)

// find
println(map3.getOrElse(key = "Alice"){ "Unknown" })


```

### Get or Put

```kotlin
val mutableMap = map3.toMutableMap()
println(mutableMap)
println(mutableMap.getOrPut(key = "Alice123", defaultValue = { Person(name="Alice123", age=32) }))
println(mutableMap)
```
