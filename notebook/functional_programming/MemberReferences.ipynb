# Member References


```kotlin
data class Person(
    val name: String,
    val age: Int,
    val male: Boolean
)

val people = listOf<Person>(
    Person(name = "Pedro", age = 25, male = true),
    Person(name = "Petra", age = 25, male = false),
    Person(name = "Jaime", age = 30, male = true),
    Person(name = "Jamie", age = 28, male = false),
    Person(name = "Perlita", age = 21, male = false)
)
/*
println("male:")
people.filter { person -> person.male }.forEach { person -> println(person.name) }
println("female:")
people.filter { person -> !person.male }.forEach {  println(it.name) }
*/

// shorter alternative
println("male:")
people.filter(Person::male).forEach { println(it.name) }
println("female:")
people.filterNot(Person::male).forEach { println(it.name) }

```

```kotlin
// sorting
println("Ascending Sort:")
people.sortedBy(Person::age).forEach { println("${it.name} age: ${it.age}") }

println("Descending Sort:")
people.sortedByDescending (Person::age).forEach { println("${it.name} age: ${it.age}") }
```

## Function References

```kotlin
// function references
fun Person.important() = this.name.startsWith(prefix = "J") && this.age > 25

println(people.any(::important2))


fun important2(person: Person) = person.name.startsWith("J") && person.age > 26
```
